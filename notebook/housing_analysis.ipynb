Research Question 1: Which NYC borough has the highest number of counted affordable housing units?
- Columns: Borough, All Counted Units
- Visualization: Bar chart comparing the five boroughs.
Research Question 2: How are affordable housing units distributed across different income levels in each borough?
- Columns: Borough, Extremely Low Income Units, Very Low Income Units, Low Income Units, Moderate Income Units, Middle Income Units, Other Income Units
- Visualization: Stacked bar chart comparing income categories across boroughs.

```python
income_cols = [
    "Extremely Low Income Units",
    "Very Low Income Units",
    "Low Income Units",
    "Moderate Income Units",
    "Middle Income Units",
    "Other Income Units"
]

income_sum = df[income_cols].fillna(0).sum(axis=1)

difference = (
    df["All Counted Units"] - income_sum
)

print("Rows in dataset:", len(df))
print("Rows where income totals match counted units:",
      (difference == 0).sum())
print("Rows where totals do not match:",
      (difference != 0).sum())
```

```python
check = df[["Project ID", "Project Name", "Borough",
            "All Counted Units"] + income_cols].copy()

check["Income Category Sum"] = df[income_cols].fillna(0).sum(axis=1)

check["Difference"] = (
    check["All Counted Units"] - check["Income Category Sum"]
)

check["Missing Income Categories"] = df[income_cols].isnull().sum(axis=1)

mismatches = check[check["Difference"] != 0]

display(mismatches)
```

```python
income_sum = df[income_cols].fillna(0).sum(axis=1)

has_total = df["All Counted Units"].notna()
has_income_data = df[income_cols].notna().any(axis=1)

checkable = has_total & has_income_data

difference = df["All Counted Units"] - income_sum

print("Records we can compare:", checkable.sum())
print("Matching records:", (difference[checkable] == 0).sum())
print("Non-matching records:", (difference[checkable] != 0).sum())
print("Records without enough information:", (~checkable).sum())
```

```python
borough_summary = (
    df.dropna(subset=["Borough", "All Counted Units"])
    .groupby("Borough")
    .agg(
        Total_Count_Units=("All Counted Units", "sum"),
        Number_of_Records=("Building ID", "count")
    )
    .sort_values("Total_Count_Units", ascending=False)
)

display(borough_summary)
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    borough_summary.index,
    borough_summary["Total_Count_Units"]
)

plt.title("Affordable Housing Units by NYC Borough")
plt.xlabel("Borough")
plt.ylabel("Total Counted Units")

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
income_cols = [
    "Extremely Low Income Units",
    "Very Low Income Units",
    "Low Income Units",
    "Moderate Income Units",
    "Middle Income Units",
    "Other Income Units"
]

# Keep records with available income information
income_data = df[
    df[income_cols].notna().any(axis=1)
].copy()

# Missing individual categories count as zero
# only in records that contain income information
income_data[income_cols] = income_data[income_cols].fillna(0)

# Calculate totals for each borough
borough_income = (
    income_data.groupby("Borough")[income_cols]
    .sum()
)

display(borough_income)
```

```python
fig, ax = plt.subplots(figsize=(13, 7))

borough_income.plot(
    kind="bar",
    stacked=True,
    ax=ax
)

ax.set_title(
    "Affordable Housing Units by Income Level and Borough",
    fontsize=14
)

ax.set_xlabel("NYC Borough")
ax.set_ylabel("Number of Housing Units")

ax.legend(
    title="Income Category",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

ax.grid(axis="y", alpha=0.3)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

## Research Findings

### Research Question 1
Which NYC borough has the highest number of counted affordable housing units?

The Bronx ranked first with 106,661 counted units, followed by Brooklyn with 90,784 units. Staten Island had the lowest total, with 6,211 units.

### Research Question 2
How are affordable housing units distributed across different income levels in each NYC borough?

Low Income Units were the largest income category in the Bronx, Brooklyn, Manhattan, and Staten Island. In Queens, Very Low Income Units were the largest category.

The results show differences in the distribution of affordable housing units across NYC boroughs.

### Data Limitation
29 records had missing values across all income categories and All Counted Units. These records were excluded from the calculations because their unit totals were unavailable.
