```python
# Make copies so original datasets stay unchanged
housing_int_clean = housing_int.copy()
crashes_int_clean = crashes_int.copy()

# Standardize borough names
housing_int_clean["BOROUGH_KEY"] = (
    housing_int_clean["Borough"]
    .astype("string")
    .str.strip()
    .str.upper()
)

crashes_int_clean["BOROUGH_KEY"] = (
    crashes_int_clean["BOROUGH"]
    .astype("string")
    .str.strip()
    .str.upper()
)

# Standardize postcodes and ZIP codes
def clean_zip(series):
    numbers = pd.to_numeric(series, errors="coerce")
    numbers = numbers.where(numbers % 1 == 0)
    return numbers.astype("Int64").astype("string")

housing_int_clean["ZIP_KEY"] = clean_zip(
    housing_int_clean["Postcode"]
)

crashes_int_clean["ZIP_KEY"] = clean_zip(
    crashes_int_clean["ZIP CODE"]
)

print("HOUSING LOCATION EXAMPLES:")
display(housing_int_clean[["BOROUGH_KEY", "ZIP_KEY"]].head(10))

print("\nCRASH LOCATION EXAMPLES:")
display(crashes_int_clean[["BOROUGH_KEY", "ZIP_KEY"]].head(10))

print("\nHOUSING records with both location keys:",
      housing_int_clean.dropna(
          subset=["BOROUGH_KEY", "ZIP_KEY"]
      ).shape[0])

print("CRASH records with both location keys:",
      crashes_int_clean.dropna(
          subset=["BOROUGH_KEY", "ZIP_KEY"]
      ).shape[0])
```

```python
housing_for_merge = housing_int_clean.dropna(
    subset=["BOROUGH_KEY", "ZIP_KEY"]
).copy()

housing_summary = (
    housing_for_merge
    .groupby(["BOROUGH_KEY", "ZIP_KEY"], as_index=False)
    .agg(
        Housing_Building_Records=("Borough", "size"),
        Housing_Total_Units=(
            "All Counted Units",
            lambda x: x.sum(min_count=1)
        ),
        Missing_Housing_Unit_Records=(
            "All Counted Units", "count"
        )
    )
)

display(housing_summary.head(10))

print("Housing geographic groups:", len(housing_summary))
```

```python
crashes_for_merge = crashes_int_clean.dropna(
    subset=["BOROUGH_KEY", "ZIP_KEY"]
).copy()

crash_summary = (
    crashes_for_merge
    .groupby(["BOROUGH_KEY", "ZIP_KEY"], as_index=False)
    .agg(
        Number_of_Crashes=("COLLISION_ID", "count"),
        Total_Persons_Injured=(
            "NUMBER OF PERSONS INJURED",
            lambda x: x.sum(min_count=1)
        ),
        Total_Persons_Killed=(
            "NUMBER OF PERSONS KILLED",
            lambda x: x.sum(min_count=1)
        )
    )
)

display(crash_summary.head(10))

print("Crash geographic groups:", len(crash_summary))
```

```python
housing_summary = (
    housing_for_merge
    .groupby(["BOROUGH_KEY", "ZIP_KEY"], as_index=False)
    .agg(
        Housing_Records=("Borough", "size"),
        Housing_Total_Units=(
            "All Counted Units",
            lambda x: x.sum(min_count=1)
        ),
        Records_With_Unit_Count=(
            "All Counted Units", "count"
        ),
        Records_Without_Unit_Count=(
            "All Counted Units",
            lambda x: x.isna().sum()
        )
    )
)

display(housing_summary.head(10))
```

```python
integrated_df = housing_summary.merge(
    crash_summary,
    on=["BOROUGH_KEY", "ZIP_KEY"],
    how="outer",
    validate="one_to_one",
    indicator=True
)

print("Integrated dataset shape:", integrated_df.shape)

print("\nIntegration results:")
print(integrated_df["_merge"].value_counts())

print("\nIntegrated dataset preview:")
display(integrated_df.head(15))
```

```python
# Count each type of integration result
print(integrated_df["_merge"].value_counts())

# Display housing-only locations
print("\nHOUSING-ONLY LOCATIONS:")
display(
    integrated_df[integrated_df["_merge"] == "left_only"]
)

# Display the first 20 crash-only locations
print("\nCRASH-ONLY LOCATIONS:")
display(
    integrated_df[integrated_df["_merge"] == "right_only"]
    .head(20)
)
```

```python
crash_only = integrated_df[
    integrated_df["_merge"] == "right_only"
]

print(
    crash_only.groupby("BOROUGH_KEY").size()
)
```

```python
crash_only = integrated_df[
    integrated_df["_merge"] == "right_only"
]

print("UNMATCHED CRASH LOCATIONS:\n")

display(
    crash_only[
        [
            "BOROUGH_KEY",
            "ZIP_KEY",
            "Number_of_Crashes"
        ]
    ].sort_values(
        ["BOROUGH_KEY", "ZIP_KEY"]
    ).to_string(index=False)
)
```

```python
# Create a separate copy
integrated_clean = integrated_df.copy()

# Give the merge results understandable names
integrated_clean["MATCH_STATUS"] = integrated_clean["_merge"].map({
    "both": "Matched",
    "left_only": "Housing Only",
    "right_only": "Crash Only"
})

# Remove the technical merge indicator
integrated_clean = integrated_clean.drop(columns=["_merge"])

# Flag the suspicious ZIP value for investigation
integrated_clean["REVIEW_NOTE"] = ""

integrated_clean.loc[
    integrated_clean["ZIP_KEY"] == "12345",
    "REVIEW_NOTE"
] = "Verify ZIP against original crash location"

# Sort the integrated data
integrated_clean = integrated_clean.sort_values(
    ["BOROUGH_KEY", "ZIP_KEY"]
).reset_index(drop=True)

print("Integrated dataset shape:", integrated_clean.shape)

print("\nIntegration status:")
print(integrated_clean["MATCH_STATUS"].value_counts())

print("\nDuplicate geographic keys:",
      integrated_clean.duplicated(
          subset=["BOROUGH_KEY", "ZIP_KEY"]
      ).sum())

display(integrated_clean.head(15))
```

```python

# ==========================================
# NYC HOUSING & CRASHES DATA INTEGRATION
# ==========================================

import pandas as pd
import numpy as np

# STEP 1: LOAD CLEANED DATASETS

housing_clean = pd.read_csv("../data/housing_clean.csv")
crashes = pd.read_csv("../data/crashes_clean.csv", low_memory=False)

print("HOUSING DATA:", housing_clean.shape)
print("CRASH DATA:", crashes.shape)


# STEP 2: STANDARDIZE BOROUGH AND ZIP CODE

# Housing dataset
housing_clean["Borough"] = (
    housing_clean["Borough"]
    .astype("string")
    .str.strip()
    .str.upper()
)

housing_clean["Postcode"] = (
    pd.to_numeric(housing_clean["Postcode"], errors="coerce")
    .astype("Int64")
    .astype("string")
    .str.zfill(5)
)

# Crashes dataset
crashes["BOROUGH"] = (
    crashes["BOROUGH"]
    .astype("string")
    .str.strip()
    .str.upper()
)

crashes["ZIP CODE"] = (
    pd.to_numeric(crashes["ZIP CODE"], errors="coerce")
    .astype("Int64")
    .astype("string")
    .str.zfill(5)
)


# STEP 3: FILTER RECORDS WITH USABLE GEOGRAPHIC KEYS

housing_geo = housing_clean.dropna(
    subset=["Borough", "Postcode"]
).copy()

crashes_geo = crashes.dropna(
    subset=["BOROUGH", "ZIP CODE"]
).copy()

print("\nHOUSING RECORDS WITH BOROUGH AND ZIP:", len(housing_geo))
print("CRASH RECORDS WITH BOROUGH AND ZIP:", len(crashes_geo))


# STEP 4: AGGREGATE HOUSING DATA BY BOROUGH AND ZIP

housing_by_zip = (
    housing_geo.groupby(["Borough", "Postcode"], as_index=False)
    .agg(
        housing_projects=("Project ID", "nunique"),
        counted_housing_units=("All Counted Units", "sum"),
        total_units=("Total Units", "sum")
    )
)

print("\nHOUSING GEOGRAPHIC GROUPS:", len(housing_by_zip))


# STEP 5: AGGREGATE CRASH DATA BY BOROUGH AND ZIP

crashes_by_zip = (
    crashes_geo.groupby(["BOROUGH", "ZIP CODE"], as_index=False)
    .agg(
        total_crashes=("COLLISION_ID", "nunique"),
        total_injured=("NUMBER OF PERSONS INJURED", "sum"),
        total_killed=("NUMBER OF PERSONS KILLED", "sum")
    )
)

# Rename columns to match housing dataset
crashes_by_zip = crashes_by_zip.rename(
    columns={
        "BOROUGH": "Borough",
        "ZIP CODE": "Postcode"
    }
)

print("CRASH GEOGRAPHIC GROUPS:", len(crashes_by_zip))


# STEP 6: MERGE BOTH DATASETS

integrated = pd.merge(
    housing_by_zip,
    crashes_by_zip,
    on=["Borough", "Postcode"],
    how="outer",
    indicator=True
)


# STEP 7: IDENTIFY MATCHED AND UNMATCHED LOCATIONS

integrated["match_status"] = integrated["_merge"].map({
    "both": "Matched",
    "left_only": "Housing only",
    "right_only": "Crashes only"
})


# STEP 8: HANDLE MISSING AGGREGATE VALUES

count_columns = [
    "housing_projects",
    "counted_housing_units",
    "total_units",
    "total_crashes",
    "total_injured",
    "total_killed"
]

integrated[count_columns] = integrated[count_columns].fillna(0)

# Remove the temporary merge indicator
integrated = integrated.drop(columns=["_merge"])


# STEP 9: DISPLAY INTEGRATION RESULTS

print("\n========== INTEGRATION RESULTS ==========")

print("Integrated dataset shape:", integrated.shape)

print("\nMATCHING STATUS:")
print(integrated["match_status"].value_counts())

print("\nFIRST 10 ROWS:")
display(integrated.head(10))


# STEP 10: SAVE THE NEW INTEGRATED DATASET

integrated.to_csv(
    "../data/integrated_clean_updated.csv",
    index=False
)

print("\nSuccessfully saved: data/integrated_clean_updated.csv")
```

Research question 5: Is there an association between the number of housing units and the number of crashes across NYC ZIP codes?

```python

import plotly.express as px

# Use only locations matched in both datasets
matched_data = integrated[
    integrated["match_status"] == "Matched"
].copy()

# Remove locations with no counted housing units
matched_data = matched_data[
    matched_data["counted_housing_units"] > 0
]

fig1 = px.scatter(
    matched_data,
    x="counted_housing_units",
    y="total_crashes",
    color="Borough",
    hover_data=[
        "Postcode",
        "housing_projects",
        "total_injured",
        "total_killed"
    ],
    title="Housing Units vs. Motor Vehicle Crashes by ZIP Code",
    labels={
        "counted_housing_units": "Counted Housing Units",
        "total_crashes": "Number of Crashes",
        "Borough": "NYC Borough"
    },
    template="plotly_white"
)

fig1.update_traces(marker=dict(size=10, opacity=0.7))

fig1.update_layout(
    title_x=0.5,
    height=600
)

fig1.show()
```

Research question 6 : How do housing development and motor vehicle crash counts compare across the five NYC boroughs?

```python

from plotly.subplots import make_subplots
import plotly.graph_objects as go

# Aggregate integrated data by borough
borough_summary = (
    integrated.groupby("Borough", as_index=False)
    .agg(
        total_housing_units=("counted_housing_units", "sum"),
        total_crashes=("total_crashes", "sum"),
        total_injured=("total_injured", "sum"),
        total_killed=("total_killed", "sum")
    )
)

# Create a chart with two y-axes
fig2 = make_subplots(specs=[[{"secondary_y": True}]])

# Housing units
fig2.add_trace(
    go.Bar(
        x=borough_summary["Borough"],
        y=borough_summary["total_housing_units"],
        name="Counted Housing Units"
    ),
    secondary_y=False
)

# Crashes
fig2.add_trace(
    go.Scatter(
        x=borough_summary["Borough"],
        y=borough_summary["total_crashes"],
        name="Total Crashes",
        mode="lines+markers",
        line=dict(width=3),
        marker=dict(size=10)
    ),
    secondary_y=True
)

fig2.update_layout(
    title="Housing Units and Motor Vehicle Crashes by NYC Borough",
    title_x=0.5,
    template="plotly_white",
    height=600,
    legend_title="Metrics"
)

fig2.update_xaxes(title_text="NYC Borough")
fig2.update_yaxes(
    title_text="Counted Housing Units",
    secondary_y=False
)
fig2.update_yaxes(
    title_text="Number of Crashes",
    secondary_y=True
)

fig2.show()

# Display the borough summary
display(borough_summary)
```

Important interpretation note: These charts show geographic associations, not proof that housing development causes crashes. The datasets may cover different time periods, and housing units are not a direct measure of population or traffic exposure.

# Data Integration and Post-Integration Cleaning

## Integration Method

The NYC Housing and Development dataset was combined with the NYC Motor Vehicle Collisions dataset.

The integration used two shared geographic attributes:
- Borough
- Postcode / ZIP Code

Both datasets were standardized to use uppercase borough names and consistent five-digit ZIP code formats.

## Data Aggregation

Housing records were grouped by borough and postcode to calculate:
- Number of housing building records
- Total counted housing units
- Total units

Crash records were grouped by borough and ZIP code to calculate:
- Number of crashes
- Total persons injured
- Total persons killed

## Integration Strategy

An outer merge was selected to preserve all geographic groups from both datasets, including locations that appeared in only one dataset.

This allowed us to identify matched locations, crash-only locations, and housing-only locations instead of losing unmatched records.

## Integration Results

The two summarized datasets were combined using an outer merge.

Results:
- Total geographic locations: 242
- Matched locations: 158
- Crash-only locations: 83
- Housing-only locations: 1
- Duplicate geographic keys: 0

## Post-Integration Cleaning

- Standardized borough names and ZIP codes before merging.
- Identified unmatched records using the merge indicator.
- Created a `match_status` column to classify geographic groups.
- Filled missing aggregated count values with zero after merging for numerical comparisons.
- Preserved missing borough and ZIP code information rather than inventing geographic values.

## Output

The integrated dataset was exported as:

`data/integrated_clean_updated.csv`

## Data Limitations

- Some locations appeared in only one dataset.
- Missing borough and ZIP code information limited geographic matching.
- Some borough and ZIP code combinations may require further verification.
- The analysis shows associations and comparisons, not evidence that housing development causes traffic crashes.
