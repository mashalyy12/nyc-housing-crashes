```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the housing data
housing_int = pd.read_csv(
    "../data/housing.csv",
    usecols=[
        "Borough",
        "Postcode",
        "All Counted Units"
    ]
)

# Load the crash data
crashes_int = pd.read_csv(
    "../data/crashes.csv",
    usecols=[
        "BOROUGH",
        "ZIP CODE",
        "COLLISION_ID",
        "NUMBER OF PERSONS INJURED",
        "NUMBER OF PERSONS KILLED"
    ],
    low_memory=False
)

print("Both datasets loaded successfully!")
```

```python
print("HOUSING DATA:")
print("Shape:", housing_int.shape)
display(housing_int.head())

print("\nCRASH DATA:")
print("Shape:", crashes_int.shape)
display(crashes_int.head())
```

```python
print("HOUSING DATA")
print("Shape:", housing_int.shape)
print("Data types:")
print(housing_int.dtypes)
print("\nMissing values:")
print(housing_int.isnull().sum())

print("\nCRASH DATA")
print("Shape:", crashes_int.shape)
print("Data types:")
print(crashes_int.dtypes)
print("\nMissing values:")
print(crashes_int.isnull().sum())

print("\nFirst 5 crash records:")
display(crashes_int.head())
```

```python
print("CRASH DATA MISSING VALUES:")
print(crashes_int.isnull().sum().to_string())

print("\nCRASH BOROUGH COUNTS:")
print(crashes_int["BOROUGH"].value_counts(dropna=False))

print("\nHOUSING BOROUGH COUNTS:")
print(housing_int["Borough"].value_counts(dropna=False))

print("\nHOUSING POSTCODE EXAMPLES:")
print(housing_int["Postcode"].dropna().head(10).tolist())

print("\nCRASH ZIP CODE EXAMPLES:")
print(crashes_int["ZIP CODE"].dropna().head(10).tolist())
```

```python
# Make copies so original datasets stay unchanged
housing_int_clean = housing_int.copy()
crashes_int_clean = crashes_int.copy()

# Standardize borough names
housing_int_clean["BOROUGH_KEY"] = (
    housing_int_clean["Borough"]
    .astype("string")
    .str.strip()
    .str.upper()
)

crashes_int_clean["BOROUGH_KEY"] = (
    crashes_int_clean["BOROUGH"]
    .astype("string")
    .str.strip()
    .str.upper()
)

# Standardize postcodes and ZIP codes
def clean_zip(series):
    numbers = pd.to_numeric(series, errors="coerce")
    numbers = numbers.where(numbers % 1 == 0)
    return numbers.astype("Int64").astype("string")

housing_int_clean["ZIP_KEY"] = clean_zip(
    housing_int_clean["Postcode"]
)

crashes_int_clean["ZIP_KEY"] = clean_zip(
    crashes_int_clean["ZIP CODE"]
)

print("HOUSING LOCATION EXAMPLES:")
display(housing_int_clean[["BOROUGH_KEY", "ZIP_KEY"]].head(10))

print("\nCRASH LOCATION EXAMPLES:")
display(crashes_int_clean[["BOROUGH_KEY", "ZIP_KEY"]].head(10))

print("\nHOUSING records with both location keys:",
      housing_int_clean.dropna(
          subset=["BOROUGH_KEY", "ZIP_KEY"]
      ).shape[0])

print("CRASH records with both location keys:",
      crashes_int_clean.dropna(
          subset=["BOROUGH_KEY", "ZIP_KEY"]
      ).shape[0])
```

```python
housing_for_merge = housing_int_clean.dropna(
    subset=["BOROUGH_KEY", "ZIP_KEY"]
).copy()

housing_summary = (
    housing_for_merge
    .groupby(["BOROUGH_KEY", "ZIP_KEY"], as_index=False)
    .agg(
        Housing_Building_Records=("Borough", "size"),
        Housing_Total_Units=(
            "All Counted Units",
            lambda x: x.sum(min_count=1)
        ),
        Missing_Housing_Unit_Records=(
            "All Counted Units", "count"
        )
    )
)

display(housing_summary.head(10))

print("Housing geographic groups:", len(housing_summary))
```

```python
crashes_for_merge = crashes_int_clean.dropna(
    subset=["BOROUGH_KEY", "ZIP_KEY"]
).copy()

crash_summary = (
    crashes_for_merge
    .groupby(["BOROUGH_KEY", "ZIP_KEY"], as_index=False)
    .agg(
        Number_of_Crashes=("COLLISION_ID", "count"),
        Total_Persons_Injured=(
            "NUMBER OF PERSONS INJURED",
            lambda x: x.sum(min_count=1)
        ),
        Total_Persons_Killed=(
            "NUMBER OF PERSONS KILLED",
            lambda x: x.sum(min_count=1)
        )
    )
)

display(crash_summary.head(10))

print("Crash geographic groups:", len(crash_summary))
```

```python
housing_summary = (
    housing_for_merge
    .groupby(["BOROUGH_KEY", "ZIP_KEY"], as_index=False)
    .agg(
        Housing_Records=("Borough", "size"),
        Housing_Total_Units=(
            "All Counted Units",
            lambda x: x.sum(min_count=1)
        ),
        Records_With_Unit_Count=(
            "All Counted Units", "count"
        ),
        Records_Without_Unit_Count=(
            "All Counted Units",
            lambda x: x.isna().sum()
        )
    )
)

display(housing_summary.head(10))
```

```python
integrated_df = housing_summary.merge(
    crash_summary,
    on=["BOROUGH_KEY", "ZIP_KEY"],
    how="outer",
    validate="one_to_one",
    indicator=True
)

print("Integrated dataset shape:", integrated_df.shape)

print("\nIntegration results:")
print(integrated_df["_merge"].value_counts())

print("\nIntegrated dataset preview:")
display(integrated_df.head(15))
```

```python
# Count each type of integration result
print(integrated_df["_merge"].value_counts())

# Display housing-only locations
print("\nHOUSING-ONLY LOCATIONS:")
display(
    integrated_df[integrated_df["_merge"] == "left_only"]
)

# Display the first 20 crash-only locations
print("\nCRASH-ONLY LOCATIONS:")
display(
    integrated_df[integrated_df["_merge"] == "right_only"]
    .head(20)
)
```

```python
crash_only = integrated_df[
    integrated_df["_merge"] == "right_only"
]

print(
    crash_only.groupby("BOROUGH_KEY").size()
)
```

```python
crash_only = integrated_df[
    integrated_df["_merge"] == "right_only"
]

print("UNMATCHED CRASH LOCATIONS:\n")

display(
    crash_only[
        [
            "BOROUGH_KEY",
            "ZIP_KEY",
            "Number_of_Crashes"
        ]
    ].sort_values(
        ["BOROUGH_KEY", "ZIP_KEY"]
    ).to_string(index=False)
)
```

```python
# Create a separate copy
integrated_clean = integrated_df.copy()

# Give the merge results understandable names
integrated_clean["MATCH_STATUS"] = integrated_clean["_merge"].map({
    "both": "Matched",
    "left_only": "Housing Only",
    "right_only": "Crash Only"
})

# Remove the technical merge indicator
integrated_clean = integrated_clean.drop(columns=["_merge"])

# Flag the suspicious ZIP value for investigation
integrated_clean["REVIEW_NOTE"] = ""

integrated_clean.loc[
    integrated_clean["ZIP_KEY"] == "12345",
    "REVIEW_NOTE"
] = "Verify ZIP against original crash location"

# Sort the integrated data
integrated_clean = integrated_clean.sort_values(
    ["BOROUGH_KEY", "ZIP_KEY"]
).reset_index(drop=True)

print("Integrated dataset shape:", integrated_clean.shape)

print("\nIntegration status:")
print(integrated_clean["MATCH_STATUS"].value_counts())

print("\nDuplicate geographic keys:",
      integrated_clean.duplicated(
          subset=["BOROUGH_KEY", "ZIP_KEY"]
      ).sum())

display(integrated_clean.head(15))
```

```python
print("FINAL VALIDATION")

print("Total locations:", len(integrated_clean))

print(
    "Duplicate locations:",
    integrated_clean.duplicated(
        subset=["BOROUGH_KEY", "ZIP_KEY"]
    ).sum()
)

print(
    "Missing borough keys:",
    integrated_clean["BOROUGH_KEY"].isna().sum()
)

print(
    "Missing ZIP keys:",
    integrated_clean["ZIP_KEY"].isna().sum()
)

print("\nIntegration status:")
print(integrated_clean["MATCH_STATUS"].value_counts())
```

```python
from pathlib import Path

output_path = Path("../data/integrated_clean.csv")

integrated_clean.to_csv(output_path, index=False)

print("Integrated dataset successfully exported!")
print("Saved at:", output_path.resolve())
```

# Data Integration and Post-Integration Cleaning

## Integration Method

The NYC Housing and Development dataset was combined with the NYC Motor Vehicle Collisions dataset.

The integration used two shared geographic attributes:
- Borough
- Postcode / ZIP Code

Both datasets were standardized to use uppercase borough names and consistent ZIP code formats.

## Data Aggregation

Housing records were grouped by borough and postcode to calculate:
- Number of housing building records
- Total counted housing units

Crash records were grouped by borough and ZIP code to calculate:
- Number of crashes
- Total persons injured
- Total persons killed

## Integration Results

The two summarized datasets were combined using an outer merge.

Results:
- Total geographic locations: 242
- Matched locations: 158
- Crash-only locations: 83
- Housing-only locations: 1
- Duplicate geographic keys: 0

## Output

The integrated dataset was exported as:

data/integrated_clean.csv

## Data Limitations

Some locations appeared in only one dataset.
Missing values were preserved instead of being replaced with zero.
Some borough and ZIP code combinations require further verification.
